# RAG Pipeline Test (BM25 + Ollama)

Dieses Notebook testet die Retrieval-Augmented Generation Pipeline mit BM25 als Retriever und Ollama als Generator. Es werden Beispiel-Queries ausgeführt.

```python
from haystack.dataclasses import Document
from src.base.pipeline import create_rag_pipeline, run_rag_pipeline

docs = [
    Document(content="Sensor S1 misst Temperatur in Raum A im Jahr 2019."),
    Document(content="Sensor S2 misst Luftfeuchtigkeit in Raum B im Jahr 2020."),
    Document(content="Sensor S3 misst Temperatur in Raum B im Jahr 2019."),
]

rag_pipe = create_rag_pipeline(docs)
```

## Beispiel-Query 1

Welche Sensoren messen Temperatur im Jahr 2019?

```python
query = "Welche Sensoren messen Temperatur im Jahr 2019?"
result = run_rag_pipeline(rag_pipe, query)
print("Antwort:", result['answer'], "Dokumente:", result["retriever"]["documents"])
```
